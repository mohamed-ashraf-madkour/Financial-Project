### Total Loan Application

```python
total_loan_application=financial['id'].count()
plt.figure(figsize=(6,3))
plt.axis('off')

plt.text(
    0.5, 0.5,
    f"{total_loan_application:,}\nTotal Loan Applications",
    ha='center',
    va='center',
    fontsize=20,
  
color='w',
    bbox=dict(boxstyle="round,pad=0.5", facecolor="#4CAF50", edgecolor="g")
)

plt.show()
```

### Bad & Good Loan Metrices

```python
Good_Metrices = financial[financial['loan_status'].isin(["Fully Paid", "Current"])]
total_loan_application = financial['id'].count()
Good_Metrices_Application = Good_Metrices['id'].count()
Good_total_amount_received = Good_Metrices['total_payment'].sum()
Good_total_funded_amount = Good_Metrices['loan_amount'].sum()
Good_loan_application_percentage = (Good_Metrices_Application / total_loan_application) * 100
Bad_Loan_Metrices=financial[financial['loan_status'].isin({"Charged Off"})]
total_loan_application = financial['id'].count()
Bad_Loan_Metrices_Application = Bad_Loan_Metrices['id'].count()
Bad_total_amount_received = Bad_Loan_Metrices['total_payment'].sum()
Bad_total_funded_amount = Bad_Loan_Metrices['loan_amount'].sum()
Bad_loan_application_percentage = (Bad_Loan_Metrices_Application / total_loan_application) * 100
good_apps = Good_Metrices_Application
bad_apps = Bad_Loan_Metrices_Application

labels = ['Good Loans', 'Bad Loans']
sizes = [good_apps, bad_apps]
colors = ['#2ECC71', '#E74C3C']   
explode = (0.08, 0)  

plt.figure(figsize=(7,7))

plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=45,
    explode=explode,
    shadow=True,
    textprops={'fontsize': 12, 'fontweight': 'bold'}
)

plt.title("Loan Quality (Good vs Bad)", fontsize=14, fontweight='bold')

plt.show()
```

### Monthly Trend By Issue Date For Total Funded Amount

```python
Monthly_Trend=financial.sort_values('issue_date').assign(month_name=lambda x:x['issue_date'].dt.strftime('%b %y')).groupby('month_name',sort=False)['loan_amount'].sum().div(1000_000)
plt.figure(figsize=(12,5))

plt.plot(
    Monthly_Trend.index,
    Monthly_Trend.values,
    marker='o',
    linewidth=3,
    markersize=8,
    color='#2ECC71'
)

plt.title("Monthly Loan Trend (in Millions)", fontsize=14, fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Loan Amount (M)")

plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.4)

plt.show()
```

### Monthly Trend By Issue Date For Total Amount Received

```python
Monthly_Received=financial.sort_values('issue_date').assign(month_name=lambda x:x['issue_date'].dt.strftime('%b %y')).groupby('month_name',sort=False)['total_payment'].sum().div(1000_000)
plt.figure(figsize=(12,5))

plt.plot(
    Monthly_Received.index,
    Monthly_Received.values,
    marker='o',
    linewidth=3,
    color='#2ECC71'
)

plt.title("Monthly Amount Received (in Millions)", fontsize=14, fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Amount Received (M)")

plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.4)

plt.show()
```

### Monthly Trend By Issue Date For Total Loan Application

```python
Monthly_Total_Loan_Application=financial.sort_values('issue_date').assign(month_name=lambda x:x['issue_date'].dt.strftime('%b %y')).groupby('month_name',sort=False)['id'].sum().div(1000_000)
Monthly_Received = financial.sort_values('issue_date') \
    .assign(month_name=lambda x: x['issue_date'].dt.strftime('%b %y')) \
    .groupby('month_name', sort=False)['total_payment'] \
    .sum() / 1_000_000  

plt.figure(figsize=(12,5))

plt.plot(
    Monthly_Received.index,
    Monthly_Received.values,
    marker='s',          
    linestyle='--',      
    linewidth=2.5,
    color='#2ECC71'
)

plt.title("Monthly Amount Received (in Millions)", fontsize=14, fontweight='bold')
plt.xlabel("Month")
plt.ylabel("Amount Received (M)")

plt.xticks(rotation=45)
plt.grid(True, linestyle=':', alpha=0.5)

plt.show()
```

### Top 10 Regional Anaysis by State for Total Funded Amount

```python
region_analysis=financial.groupby('address_state')['loan_amount'].sum().nlargest(10).sort_values()
plt.figure(figsize=(10,6))

plt.barh(
    region_analysis.index,
    region_analysis.values,
    color='#2ECC71'
)

plt.title("Top 10 States by Loan Amount", fontsize=14, fontweight='bold')
plt.xlabel("Loan Amount")

plt.grid(axis='x', linestyle='--', alpha=0.4)

plt.show()
```

### Loan Term Analysis by total Funded Amount

```python
term_analysis=financial.groupby('term')['loan_amount'].sum()
plt.figure(figsize=(6,6))
vcolors = ['#2ECC71', '#E74C3C']   
plt.pie(
    term_analysis.values,
    labels=term_analysis.index,
    autopct='%1.1f%%',
    startangle=60,
    colors=vcolors,
    explode=[0.05]*len(term_analysis)
)

plt.title("Loan Amount Distribution by Term", fontsize=14, fontweight='bold')

plt.show()
```

### Employee Length by total Funded Amount 

```python
employee_analysis=financial.groupby('emp_length')['loan_amount'].sum().sort_values()
plt.figure(figsize=(10,5))

plt.bar(
    employee_analysis.index.astype(str),
    employee_analysis.values,
    color='#2ECC71'
)

plt.title("Loan Amount by Employee Length", fontsize=14, fontweight='bold')
plt.xlabel("Employment Length")
plt.ylabel("Loan Amount")

plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.4)

plt.show()
```
